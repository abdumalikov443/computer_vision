**OpenCv video streaming codes**

```python
print("Hello World")

import cv2

cap = cv2.VideoCapture(0) # 0 --> qaysi / nechinchi kamerani ishlatish ni bildiradi

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

while True:
    success, frame = cap.read()

    if not success:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray, 1.1, 4) # 4 --> yuzni 4 marta check qiladi
                    # 1.1 --> real vaqtda odam yuzini aniqlayotgan vaqtda 10% ga kichiraytirib o'qishni tezlashtrib beradi

    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    cv2.imshow("Real-time face Detection: ", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()


# ====================================================


# import cv2

# url = "rtsp://192.168.1.164:8001/video" # stream_path

# cap = cv2.VideoCapture(url)

# while True:
#     success, frame = cap.read() # True, False | Boolean

#     if not success:
#         break

#     gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
#     blurred = cv2.GaussianBlur(frame, (15, 15), 0)
#     edge1 = cv2.Canny(frame, 100, 200) # edge detection
#     edge2 = cv2.Canny(frame, 50, 250)

#     flipped = cv2.flip(frame, 1)

#     cv2.imshow("Original: ", frame)
#     # cv2.imshow("Gray: ", gray)
#     # cv2.imshow("Blurred: ", blurred) # Noise

#     # cv2.imshow("edge1: ", edge1)
#     # cv2.imshow("edge2: ", edge2)

#     cv2.imshow("flipped: ", flipped)


#     if cv2.waitKey(1) & 0xFF == ord('q'):
#         break

# cap.release()
# cv2.destroyAllWindows()
```
