```python
from bing_image_downloader import downloader
```

```python
downloader.download('uzbek pilaf plov palov food',
                    limit=100,
                    output_dir='/content/pilaf_images',
                    adult_filter_off=True, # hech qanaqa filterlarsiz download qilib beradi
                    force_replace=False, # agar output_dir bizda oldindan bor bo'lsa yaratib o'tirmaydi
                    timeout=60  # download vaqtida muammo chiqsa 60 sekund kutadida keyin boshqa rasmga o'tib ketadi
                    )
```

```python
import os, shutil

source_dir = '/content/pilaf_images/uzbek pilaf plov palov food'
target_dir = '/content/drive/MyDrive/food101_dataset/pilaf'

os.makedirs(target_dir, exist_ok=True)

for filename in os.listdir(source_dir):
  src_file = os.path.join(source_dir, filename)
  dst_file = os.path.join(target_dir, filename)
  shutil.move(src_file, dst_file)
```
